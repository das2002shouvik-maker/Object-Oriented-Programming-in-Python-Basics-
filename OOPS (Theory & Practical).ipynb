```python
class Calculator:
    def add(self,a,b): # Instance Method
        print(a+b)
    def diff(self,a,b): # Instance Method
        print(a-b)
    def multi(self,a,b): # Instance Method
        print(a*b)
        
calc=Calculator() # Reference variable or the object.

calc.add(10,20)
calc.add(30,40)
calc.diff(40,20)
calc.multi(40,10)
```

```python
# write a python program to create a student class and creates an object to it, call the method talk().

class Student:
    def __init__(self,name,age,marks): # Constructor (__init__)
        self.name=name # Instance Variable
        self.age=age
        self.marks=marks
    def talk(self): # Instance Method
        print("Your name is:",self.name)
        print("Your age is:",self.age)
        print("Your marks is:",self.marks)

s1=Student("Shouvik",24,89.67) # Reference variable
s1.talk()   
```

## Instance Variable

```python
# Instance Variable : If the value of the variable is varied again and again with every objects created then such type of a variable
# is called an instance variable.
# For every object separate copy of instance variable is created.

class Employee:
    def __init__(self): # Constructor
        self.a=40 # Instance Variable
    def m1(self): # Instance Method
        self.c=60
e=Employee()
print(e.a)
e.b=90
print(e.b)
e.m1()
print(e.c)
print(e.__dict__)
print(e.a,e.b,e.c)
```

```python
# To delete an instance variable inside and outside the class :

class Test:
    def __init__(self): # Constructor
        self.a=10 # Instance Variable
        self.b=20
        self.c=30
        self.d=40
        self.e=50
    def m1(self): # Instance Method
        del self.b

t=Test()
print(t.__dict__)
t.m1()
print(t.__dict__)
del t.d # Deleting outside the class
print(t.__dict__)
t2=Test()
print(t2.__dict__)
print(t2.d)
```

## Static Variable

```python
# Static Variable : A variable in which the value is consistent over all the objects of the class and doesnot vary from
# object to object is called static variable.
# It is declared inside the class and outside the object.
# For every object the same copy of the variable is shared among them.

class Test:
    x = 10  # Static Variable
    def __init__(self): # Constructor
        self.b=40 # Instance Variable
    def m1(self): # Instance Method
        self.g=66

t=Test()
print(t.x)
print(t.b)
t.m1()
print(t.g)
print(t.x,t.b,t.g)
print(Test.__dict__)
```

```python
class Test:
    a=10 # Static / Class variable
    def __init__(self): # Constructor
        Test.b=20 # Static Variable
    def m1(self): # Static Method
        Test.c=30 
    @classmethod # Decorator
    def m2(cls): # Class Method
        cls.d=40
        Test.e=50
    @staticmethod # Decorator
    def m3():
        Test.f=60 

t=Test()
# print(Test.__dict__)
t.m1()
t.m2()
t.m3()
Test.g=70
print(Test.__dict__)
```

```python
class Test:
    a=10 # Static Variable
    b=20
    c=30
    d=40
    e=50
    f=60
    def __init__(self): # Constructor
        print(self.a)
        print(Test.a)
    def m1(self): # Instance Method
        print(self.b)
        print(Test.b)
    @classmethod
    def m2(cls): # Class Method
        print(cls.c)
        print(Test.c)
    @staticmethod
    def m3(): # Static Method
        print(Test.d)

t=Test()
t.m1()
t.m2()
t.m3()
print(t.e)
print(Test.__dict__)
```

# Methods in python class:

## Instance Method

```python
# Using instance method we can access the instance variales using the self parameter.

# A program to find the grade of a particular student.

class Student:
    def __init__(self,name,rollno,marks): # Constructor
        self.name=name # Instance Variable
        self.rollno=rollno
        self.marks=marks
    def display(self): # Instance Method
        print("Your name is:",self.name)
        print("Your roll no. is:",self.rollno)
        print("Marks obtained is:",self.marks)
    def grade(self): # Instance Method
        if (self.marks>80):
            print("Grade A")
        elif (self.marks>65):
            print("Grade B")
        elif (self.marks>50):
            print("Grade C")
        elif (self.marks>35):
            print("Grade D")
        else:
            print("Grade E")

s1=Student("Shouvik",28,99) # Reference Variable / Reference Object
s1.display() # Calling method display()
s1.grade() # Calling method grade()
```

```python
# Setter and Getter Method:

# The setter method is used to set the value of the instance variable and the getter method is used to get the value of the
# instance varaible. Setter method is also known as mutator method and getter method is known as accessor method.

class Test():
    def setage(self,age): # Setter Method
        self.age=age
    def getage(self): # Getter method
        return self.age

t=Test()
t.setage(24)
t.getage()
```

## Class Method

```python


class animal:
    legs=4 # Static Variable
    @classmethod # Decorator
    def walk(cls,name): # name --> local variable
        print("A",name,"has",cls.legs,"legs")

a=animal()
a.walk("Dog")
```

## Static Method

```python
# These are general utility methods.

class Test:
    @staticmethod 
    def add(a,b): # Static Method
        print(a+b)

t=Test()
t.add(10,20)
```

```python
# Write a program to define the class as company and its employee, manager, employee salary, turnover of the company.

class Company:
    mgr_name="Rahul" # Static Variable / Class Variable
    mgr_office="Pune"
    def __init__(self,empname,empsalary): # Constructor
        self.empname=empname # Instance Variable
        self.empsalary=empsalary
    def empdetails(self): # Instance Method
        print("Employee Name:",self.empname)
        print("Employee Salary:",self.empsalary)
    @classmethod
    def mgrdetails(cls,office_no): # Class Method
        print("The manager of the company is:",cls.mgr_name) 
        print("The office of the manager is in:",cls.mgr_office) # Static Variable
        print("The office number of the manager is:",office_no) # Local Variable
    @staticmethod
    def cturnover(a): # Static Method
        print("The turnover for the company is:",a,"Rs.")

e=Company("Shouvik",50000)
e.empdetails()
e.mgrdetails(2)
e.cturnover(20000000)
```
